```python
import os
import splitfolders
from sklearn.metrics import classification_report, confusion_matrix

# Path to Folders
initial_folder = '/Users/purnaperamune/Desktop/SafeMush/MushroomDatasetSample'
dataset = '/Users/purnaperamune/Desktop/SafeMush/Mushroom_Data'

splitfolders.ratio(initial_folder, output=dataset, seed=42, ratio=(.7, .15, .15))

def count_files(directory):
    total_count = 0
    for root, dirs, files in os.walk(directory):
        total_count += len([file for file in files if file.lower().endswith(('.png'))])
    return total_count

# Datasets for training, validation and testing
train_path = os.path.join(dataset, 'train')
val_path = os.path.join(dataset, 'val')
test_path = os.path.join(dataset, 'test')

num_train = count_files(train_path)
num_val = count_files(val_path)
num_test = count_files(test_path)

print(f"Training set: {num_train} images")
print(f"Validation set: {num_val} images")
print(f"Testing set: {num_test} images")
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
base_dir = '/Users/purnaperamune/Desktop/SafeMush/Mushroom_Data'
# Applying Image augementation techqniques on the training dataset
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest')

validation_datagen = ImageDataGenerator(rescale=1./255)

# Testing dataset will not be augemented, but only rescaled.
test_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(
    base_dir + '/train',
    target_size=(224, 224),
    batch_size=8,
    class_mode='binary')

validation_generator = validation_datagen.flow_from_directory(
    base_dir + '/val',
    target_size=(224, 224),
    batch_size=8,
    class_mode='binary')

test_generator = test_datagen.flow_from_directory(
    base_dir + '/test',
    target_size=(224, 224),
    batch_size=8,
    class_mode='binary',
    shuffle=False
) 
```

```python
#### {'edible': 0, 'poisonous': 1}

#In the context of your ImageDataGenerator and the flow_from_directory method, the classes (edible and poisonous) are automatically assigned labels based on the alphabetical order of the folders in your directory. Given your folder structure, where "edible" comes before "poisonous" alphabetically, the class "edible" would be assigned the label 0, and "poisonous" would be assigned the label 1.
#abels array: The sequence of 0s followed by 1s in your labels array (y_test) indicates that the first portion of your batch consists of images of edible mushrooms, and the second portion consists of images of poisonous mushrooms. This is expected behavior when shuffle=False is set in your test_generator. If shuffle were set to True, these labels would be mixed.
```

```python
X_train, y_train = train_generator.next()
print(f'This is the X_train: Batch shape=%s, min=%.3f, max=%.3f' % (X_train.shape, X_train.min(), X_train.max()))

X_validation, y_validation = validation_generator.next()
print(f'This is the X_validation: Batch shape=%s, min=%.3f, max=%.3f' % (X_validation.shape, X_validation.min(), X_validation.max()))


X_test, y_test = test_generator.next()
print(f'This is the X_test: Batch shape=%s, min=%.3f, max=%.3f' % (X_test.shape, X_test.min(), X_test.max()))
print(y_test)
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_images(generator, title):
    # Generate a batch of images
    x, y = next(generator)
    
    # Plot the images in the batch, along with their labels
    plt.figure(figsize=(10, 10))
    for i in range(8): # Display the first 9 images and labels from the batch
        plt.subplot(3, 3, i + 1)
        plt.imshow(x[i])
        plt.title('Edible' if y[i] == 0 else 'Poisonous')
        plt.axis('off')
    plt.suptitle(title)
    plt.show()

# Visualize training, validation, and test images
plot_images(train_generator, 'Training Images')
plot_images(validation_generator, 'Validation Images')
plot_images(test_generator, 'Test Images')
```

# MODEL IMPLEMENTATION

```python
from tensorflow.keras.layers import Input, Concatenate, Conv2D, BatchNormalization, Activation, GlobalAveragePooling2D, Dense
from tensorflow.keras.models import Model
from tensorflow.keras.applications import MobileNet
from tensorflow.keras.optimizers import Adam


def build_model(input_shape=(224, 224, 3)):
    # Load the base MobileNet model
    base_model = MobileNet(weights='imagenet', include_top=False, input_shape=input_shape)
    base_model.trainable = False

    # Original MobileNet features
    base_features = base_model.output
    base_features = GlobalAveragePooling2D()(base_features)
    
    # New feature branch
    new_features = Conv2D(512, (3, 3), activation=None, padding='same')(base_model.output)
    new_features = BatchNormalization()(new_features)
    new_features = Activation('relu')(new_features)
    new_features = GlobalAveragePooling2D()(new_features)
    
    # Combine features
    combined_features = Concatenate()([base_features, new_features])
    
    # Fully connected layers
    x = Dense(1024, activation='relu')(combined_features)
    predictions = Dense(1, activation='sigmoid')(x)
    
    model = Model(inputs=base_model.input, outputs=predictions)
    return model


```

```python
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

# Early stopping
early_stopping = EarlyStopping(monitor='val_loss', patience=10, verbose=1, mode='min', restore_best_weights=True)

# Learning rate schedule
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=5, verbose=1, mode='min', min_lr=1e-6)


model = build_model()
model.compile(optimizer=Adam(learning_rate=0.0001), loss='binary_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
#from tensorflow.keras.optimizers import Adam

#model.compile(optimizer=Adam(lr=0.0001), loss='binary_crossentropy', metrics=['accuracy'])
```

```python
history = model.fit(
    train_generator,
    steps_per_epoch=295,  # Number of training images // batch size
    epochs=18,
    validation_data=validation_generator,
    validation_steps= 63  # Number of validation images // batch size
)
```

# PERFORMANCE MATRIXES

```python
eval_result = model.evaluate(test_generator, steps=np.ceil(test_generator.samples/test_generator.batch_size))
print(f"Test loss: {eval_result[0]} - Test accuracy: {eval_result[1]}")
```

```python
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Make sure test_generator is not shuffled and reinitialized before predictions
test_generator.reset()

# Predict on the entire test set
predictions = model.predict(test_generator, steps=np.ceil(test_generator.samples/test_generator.batch_size))
predicted_classes = np.where(predictions > 0.5, 1, 0).flatten()

# True labels
true_classes = test_generator.classes

# Confusion Matrix
conf_matrix = confusion_matrix(true_classes, predicted_classes)
print("Confusion Matrix:")
print(conf_matrix)

# Classification Report
cls_report = classification_report(true_classes, predicted_classes, target_names=test_generator.class_indices.keys())
print("Classification Report:")
print(cls_report)

# Accuracy
accuracy = accuracy_score(true_classes, predicted_classes)
print(f"Accuracy: {accuracy:.4f}")
```

```python
import matplotlib.pyplot as plt

# Assuming 'history' is the return value from the model.fit() call
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
loss = history.history['loss']
val_loss = history.history['val_loss']

epochs = range(1, len(acc) + 1)

# Plot training and validation accuracy
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(epochs, acc, label='Training Accuracy')
plt.plot(epochs, val_acc, label='Validation Accuracy')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

# Plot training and validation loss
plt.subplot(1, 2, 2)
plt.plot(epochs, loss, label='Training Loss')
plt.plot(epochs, val_loss, label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

# PERFORMANCE ON TEST DATA

```python
from tensorflow.keras.preprocessing import image
import matplotlib.pyplot as plt

def predict_and_display_image(img_path, model):
    # Load and preprocess the image
    img = image.load_img(img_path, target_size=(224, 224))
    img_array = image.img_to_array(img)
    img_array_expanded_dims = np.expand_dims(img_array, axis=0)
    img_preprocessed = img_array_expanded_dims / 255.0

    # Predict the class
    prediction = model.predict(img_preprocessed)
    predicted_class = 'Poisonous' if prediction.flatten()[0] > 0.5 else 'Edible'

    # Display the image and the prediction
    plt.imshow(img)
    plt.title(f'Predicted Class: {predicted_class}')
    plt.axis('off')
    plt.show()

# Example usage:
test_image_path = '/Users/purnaperamune/Desktop/SafeMush/Mushroom_Data/test/Edible/edible_2228.png'
predict_and_display_image(test_image_path, model)
```

```python
import numpy as np

# Generate predictions for all images in the test set
predictions = model.predict(test_generator, steps=np.ceil(test_generator.samples/test_generator.batch_size))

predicted_classes = np.where(predictions > 0.5, 1, 0)

# Retrieing the true class labels
true_classes = test_generator.classes

# Calculating the accuracy
accuracy = np.mean(predicted_classes.flatten() == true_classes)

print(f"Test Accuracy: {accuracy:.4f}")

# Prediction for each of the image
filenames = test_generator.filenames
for i, fname in enumerate(filenames):
    print(f"Image: {fname}, Prediction: {predicted_classes[i][0]}, Actual: {true_classes[i]}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix
import seaborn as sns

# Assuming true_classes and predicted_classes are defined as follows:
# true_classes = test_generator.classes  # Actual labels
# predicted_classes = np.where(predictions > 0.5, 1, 0).flatten()  # Predicted labels from your model

# Calculate the confusion matrix
cm = confusion_matrix(true_classes, predicted_classes)

# Define the labels for the matrix
group_names = ['Actually Not Poisonous', 'Actually Poisonous']
group_counts = ["{0:0.0f}".format(value) for value in cm.flatten()]
group_percentages = ["{0:.2%}".format(value) for value in cm.flatten()/np.sum(cm)]
labels = [f"{v1}\n{v2}\n{v3}" for v1, v2, v3 in zip(group_names*2, group_counts, group_percentages)]
labels = np.asarray(labels).reshape(2,2)

categories = ['Not Poisonous', 'Poisonous']

# Plotting
fig, ax = plt.subplots()
sns.heatmap(cm, annot=labels, fmt='', cmap='Blues', ax=ax, xticklabels=categories, yticklabels=categories)

# Labeling the plot
ax.set_xlabel('Predicted labels')
ax.set_ylabel('True labels')
ax.set_title('Confusion Matrix')
plt.show()
```
